```python
import gradio as gr
import pandas as pd
import joblib

model = joblib.load("cancer_model.pkl")

FEATURES = {
    "radius_mean": 14.0, "texture_mean": 19.0, "perimeter_mean": 92.0, "area_mean": 655.0,
    "smoothness_mean": 0.096, "compactness_mean": 0.104, "concavity_mean": 0.089,
    "concave_points_mean": 0.049, "symmetry_mean": 0.181, "fractal_dimension_mean": 0.063,
}

def predict(*values):
    row = pd.DataFrame([dict(zip(FEATURES.keys(), values))])
    p = model.predict_proba(row)[0]
    return {"Malignant": float(p[0]), "Benign": float(p[1])}

demo = gr.Interface(
    fn=predict,
    inputs=[gr.Number(label=name, value=v) for name, v in FEATURES.items()],
    outputs=gr.Label(label="Prediction"),
    title="Breast Cancer - Logistic Regression",
)

demo.launch()
```
