```python
import gradio as gr
import pandas as pd
import joblib

model = joblib.load("bank_model.pkl")

def predict(age, job, marital, education, balance, housing, loan, contact, campaign, previous, poutcome):
    row = pd.DataFrame([{
        "age": age, "balance": balance, "campaign": campaign, "previous": previous,
        "job": job, "marital": marital, "education": education,
        "housing": housing, "loan": loan, "contact": contact, "poutcome": poutcome,
    }])
    p = model.predict_proba(row)[0]
    return {"NO - will not subscribe": float(p[0]), "YES - will subscribe": float(p[1])}

demo = gr.Interface(
    fn=predict,
    inputs=[
        gr.Number(label="age", value=40),
        gr.Dropdown(["admin.", "blue-collar", "entrepreneur", "housemaid", "management", "retired",
                     "self-employed", "services", "student", "technician", "unemployed", "unknown"],
                    label="job", value="management"),
        gr.Dropdown(["married", "single", "divorced"], label="marital", value="married"),
        gr.Dropdown(["primary", "secondary", "tertiary", "unknown"], label="education", value="secondary"),
        gr.Number(label="balance", value=1000),
        gr.Dropdown(["yes", "no"], label="housing loan", value="no"),
        gr.Dropdown(["yes", "no"], label="personal loan", value="no"),
        gr.Dropdown(["cellular", "telephone", "unknown"], label="contact", value="cellular"),
        gr.Number(label="campaign (contacts this campaign)", value=2),
        gr.Number(label="previous (contacts before)", value=0),
        gr.Dropdown(["failure", "other", "success", "unknown"], label="poutcome", value="unknown"),
    ],
    outputs=gr.Label(label="Prediction"),
    title="Bank Marketing - Logistic Regression",
)

demo.launch()
```
